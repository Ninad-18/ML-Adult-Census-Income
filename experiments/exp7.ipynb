```python
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
data_path = os.getenv("DATA_PATH")
```

```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv(data_path)

df.replace("?", np.nan, inplace=True)

for col in ['workclass', 'occupation', 'native.country']:
    df[col].fillna(df[col].mode()[0], inplace=True)

df.drop_duplicates(inplace=True)
```

Part 1: Independent Two-Sample T-Test

🎯 Hypothesis:

Does mean age differ between income groups?

H₀:

Mean age of <=50K and >50K groups is equal.

H₁:

Mean age differs between the two groups.

```python
# Separate age based on income group
age_low = df[df['income'] == '<=50K']['age']
age_high = df[df['income'] == '>50K']['age']

t_stat, p_value_t = stats.ttest_ind(age_low, age_high)

print("----- Independent T-Test (Age vs Income) -----")
print("T-statistic:", t_stat)
print("P-value:", p_value_t)

if p_value_t < 0.05:
    print("Result: Reject Null Hypothesis (Significant difference in mean age)")
else:
    print("Result: Fail to Reject Null Hypothesis")
```

✅ Interpretation

If:

p-value < 0.05 → Significant difference in mean age

p-value > 0.05 → No significant difference

✅ Part 2: Chi-Square Test (Categorical Association)

We test:

Is education associated with income?

H₀:

Education and income are independent.

H₁:

Education and income are related.

```python
# Create contingency table
contingency_table = pd.crosstab(df['education'], df['income'])

chi2_stat, p_value_chi, dof, expected = chi2_contingency(contingency_table)

print("\n----- Chi-Square Test (Education vs Income) -----")
print("Chi-square statistic:", chi2_stat)
print("P-value:", p_value_chi)

if p_value_chi < 0.05:
    print("Result: Reject Null Hypothesis (Education and Income are related)")
else:
    print("Result: Fail to Reject Null Hypothesis")
```

✅ Interpretation

If:

p < 0.05 → Education and income are significantly related

p > 0.05 → No significant relationship

“Why is p-value 0?”

Answer:

It is not exactly zero. It is extremely small and rounded to 0.0 due to floating-point precision. It indicates very strong statistical significance.
