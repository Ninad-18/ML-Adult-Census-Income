```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
data_path = os.getenv("DATA_PATH")
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
df = pd.read_csv(data_path)

df.replace("?", np.nan, inplace=True)

for col in ['workclass', 'occupation', 'native.country']:
    df[col].fillna(df[col].mode()[0], inplace=True)

df.drop_duplicates(inplace=True)
```

```python
numerical_cols = ['age', 'education.num', 
                  'capital.gain', 'capital.loss', 
                  'hours.per.week']

X_num = df[numerical_cols]
```

```python
scaler_std = StandardScaler()
X_standardized = scaler_std.fit_transform(X_num)

# Convert back to DataFrame
df_standardized = pd.DataFrame(X_standardized, columns=numerical_cols)

print(df_standardized.head())
```

```python
scaler_mm = MinMaxScaler()
X_normalized = scaler_mm.fit_transform(X_num)

df_normalized = pd.DataFrame(X_normalized, columns=numerical_cols)

print(df_normalized.head())
```
